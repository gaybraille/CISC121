```python
from collections import deque
```

```python
A = [3, 8, 9, 7, 6]
K = 3
```

```python
def solution(A = [], K = 0):
  for i in range(K):
    value = A.pop()       #Holds the final value of A once it is removed
    A = deque(A)          #Converts A to a double_ended queue
    A.appendleft(value)   #Appends the held value to index 0 of A
    A = list(A)           #Converts A back to a list
  return(A)
```

```python
solution(A, K)
```
