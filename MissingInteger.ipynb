```python
A = [1, 3, 6, 4, 1, 2]
```

```python
def insertion_sort(A = []):
  #Scan each value of the list
  for scan in range(len(A) - 1):
    #Compare two adjacent values: should they be swapped?
    for compare in range(scan, -1, -1):
      #Check whether right value > left value
      if (A[compare + 1] < A[compare]):
        #Swap values
        x = A[compare + 1]
        A[compare + 1] = A[compare]
        A[compare] = x
      else:
        break
```

```python
def solution(A = []):
  insertion_sort(A)
  for scan in range(len(A)):
    #If the lowest value in the list is > 1, the lowest omitted value is 1
    if (A[0] > 1):
      return 1
    for check in range(scan + 1, len(A)):
      #Checks if any values are equal to the compare value +1
      if (A[check] <= A[scan] + 1):
        break
      else:
        #If not, then that is the smallest integer not in the list
        return A[scan] + 1
  #If the smallest value is > than the highest value in the list
  return A[scan] + 1
```

```python
solution(A)
```
