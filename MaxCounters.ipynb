```python
A = [3, 4, 4, 6, 1, 4, 4]
N = 5
```

```python
def make_zero_list(N = 0):
  A = []

  for index in range(N):
    A.append(0)

  return(A)
```

```python
def solution(N = 0, A = []):
  #Set B to an empty list of length N
  B = make_zero_list(N)
  for K in range(len(A)):
    #If the value of A at K is greater than or equal to N
    if (A[K] <= N):
      #Increase the value of the index of A at K of B by one
      B[A[K] - 1] += 1
    else:
      #Set the maximum value to the minimum possible value
      max = 0
      for scan in range(len(B)):
        #If the scanned value is greater than the max value
        if (B[scan] > max):
          #Set max to the scanned value
          max = B[scan]
      for p in range (len(B)):
        #Set all of the list's values equal to the final maximum value
        B[p] = max
  return B
```

```python
solution(N, A)
```
