```python
X = 10
Y = 85
D = 30
```

```python
def solution(X = 0, Y = 0, D = 0):
  num_loops = 0       #Creates variable num_loops
  while(X < Y):
    X += D            #Increase distance travelled X by movement value D
    num_loops += 1    #Increase the number of loops by one
  return num_loops
```

```python
solution(X, Y, D)
```
