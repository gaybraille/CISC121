```python
A = [0,1,0,1,1]
```

```python
def solution(A = []):
  #Create variable to measure number of passes
  num_passing = 0
  for sel in range(len(A)):
    #If the current car isn't going east, skip it
    if(A[sel] != 0):
      continue
    #Compare all indices greater than the current value
    for comparison in range(sel + 1, len(A)):
      #If the compared car is travelling opposite to the current car
      if(A[comparison] != A[sel]):
        num_passing += 1
  return num_passing
```

```python
solution(A)
```
